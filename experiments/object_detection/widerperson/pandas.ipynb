```python
import pandas as pd

# CSV dosyalarını yükle
df1 = pd.read_csv('/home/ika1/yzlm/Re-id/object_detection_Re-ID/WiderPersonYOLO/widerperson_pytorch_results_new_global_ranking.csv')
df2 = pd.read_csv('/home/ika1/yzlm/Re-id/object_detection_Re-ID/WiderPersonYOLO/pytorch_vs_tensorrt_comparison.csv')

# İlk DataFrame'de normalize model adı oluştur
df1['Normalized_Model'] = (df1['Version'] + df1['Size']).str.upper()

# İkinci DataFrame'de normalize model adı oluştur
df2['Normalized_Model'] = df2['Model'].str.upper()

# Gerekli sütunları seç ve birleştir
merged = pd.merge(
    df1[['Normalized_Model', 'mAP@50']],
    df2[['Normalized_Model', 'TRT_FPS']],
    on='Normalized_Model',
    how='inner'
)

# Performans skorunu hesapla
merged['Score'] = merged['mAP@50'] * merged['TRT_FPS']

# Sonuçları skora göre sırala
result = merged.sort_values('Score', ascending=False).reset_index(drop=True)

# Orjinal model adlarını ekle
model_mapping = df2[['Model', 'Normalized_Model']].drop_duplicates()
result = pd.merge(result, model_mapping, on='Normalized_Model')

# Gerekli sütunları seç
result = result[['Model', 'mAP@50', 'TRT_FPS', 'Score']]

result.to_csv("best_model_scores.csv")
```
